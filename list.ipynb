```python
#Sum of List Elements

Question 1
A teacher asks students to write a program that takes multiple numbers from the user, stores them in a list, and calculates the total sum of all the elements. This helps in understanding how to handle lists and perform aggregation operations like summation.
Task
Write a program that:
Reads an integer N (the number of elements in the list).
Reads N space-separated integers into a list.
Calculates and prints the sum of the list elements.
Input Format
The first line contains an integer N, the size of the list.
The second line contains N space-separated integers.
Output Format
A single integer representing the sum of all elements in the list.
```

```python
n = int(input())

nums = []

for i in range(n):
    nums.append(int(input()))

print(sum(nums))
```

```python
#Reverse a List

Python
Question 4
A data analyst is working with a sequence of values but needs to view them in reverse order for better comparison. To achieve this, they require a program that takes a list of numbers and prints the elements from the last to the first.
```

```python
n = int(input())

nums = list(map(int, input().split()))

nums.reverse()

for num in nums:
    print(num, end=" ")
```

```python
#Remove Duplicates

Python
Question 5
Scenario
A teacher asks students to clean up a dataset of numbers by removing duplicate entries while maintaining the original order. For example, if the same number appears multiple times, it should only be kept once, in the order it first appeared. This task helps in learning about list operations and uniqueness handling.
```

```python
n = int(input())

nums = list(map(int, input().split()))

result = []

for num in nums:

    if num not in result:
        result.append(num)

print(*result)
```

```python
#Frequency of Elements


Question 6
A shopkeeper wants to analyze customer purchase data. He needs to know how many times each item (represented by numbers) appears in the list of transactions. To solve this, we must count the frequency of each element in a list while maintaining the order of their first appearance.
```

```python
n=int(input())
nums=list(map(int,input().split()))
result=[]
for num in nums:
    if num not in result:
        print(num,":", nums.count(num), sep="")
        result.append(num)
```

```python
#List Rotation

Python
Question 7
In a game, players stand in a line. After each round, the line is rotated so that the last K players move to the front of the line. To simulate this, we need a program that rotates the elements of a list to the right by K positions.
Task
```

```python
n = int(input())

nums = list(map(int, input().split()))

k = int(input())

nums = nums[-k:] + nums[:-k]

print(*nums)
```

```python
#Even and Odd S

Python
Question 8
A mathematics teacher gives students a list of numbers and asks them to separate the even numbers (divisible by 2) and the odd numbers (not divisible by 2). This helps in learning list processing and classification.
```

```python
n=int(input())
nums=list(map(int, input().split()))
even=[]
odd=[]
for i in nums:
    if i%2==0:
        even.append(i)
        
    else:
        odd.append(i)
        
        
        
print("Even:",*even)
print("Odd:",*odd)
            
```

```python
n = int(input())

nums = list(map(int, input().split()))

total = 0

for i in range(0, n, 2):
    total += nums[i]

print(total)
```
