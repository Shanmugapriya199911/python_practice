```python
Tuple Creation

Python
Question 1
A data analyst needs to ensure that user inputs are stored in an immutable data structure to prevent accidental modification. Since Python provides tuples as immutable sequences, the analyst wants to read a series of integers from the user and store them in a tuple for safe handling.
```

```python
n=int(input())
nums=tuple(map(int,input().split()))
print(nums)
```

```python
#Access Tuple Elements

Python
Question 2
A software developer wants to quickly access the boundary elements of a dataset that is stored in an immutable data structure. Since tuples are ideal for such use, the task is to read integers into a tuple and display only the first and last elements.
```

```python
n = int(input())

nums = tuple(map(int, input().split()))

print("First:", nums[0])
print("Last:", nums[-1])
```

```python
Tuple Slicing

Python
Question 3
A data scientist wants to extract a portion of data from the middle of a dataset stored in a tuple. Since tuples allow slicing, the task is to read integers into a tuple and print the slice starting from index 1 up to index -1 (excluding the last element).
```

```python
n = int(input())

nums = tuple(map(int, input().split()))

print(nums[1:-1])
```

```python
#Concatenate Tuples

Python
Question 4
In many applications, data may be collected from multiple sources and needs to be merged for further processing. Tuples, being immutable collections, can be concatenated to form a single dataset. The task is to read two tuples and concatenate them into one.
```

```python
n1 = int(input())
t1 = tuple(map(int, input().split()))

n2 = int(input())
t2 = tuple(map(int, input().split()))

result = t1 + t2

print(result)
```

```python
#Repeat Tuple

Python
Question 5
In certain applications, data may need to be repeated for simulations, testing, or cyclic patterns. Tuples in Python can be multiplied by an integer to repeat their contents. The task is to read a tuple and an integer k, then repeat the tuple k times.
```

```python
n = int(input())

t = tuple(map(int, input().split()))

k = int(input())

print(t * k)
```

```python
Check Element 
Question 6
In many real-world applications, it is important to verify whether a particular data point exists in a dataset. Since tuples are immutable sequences, searching for an element in them can be done efficiently. The task is to check whether a given element exists in a tuple and display an appropriate message.
```

```python
Nested Tuple

Python
Question 8
Sometimes, data needs to be grouped into separate categories while still being treated as a single entity. In Python, this can be achieved using nested tuples, where one tuple contains multiple sub-tuples. The task is to read two tuples and combine them into a single nested tuple.
```

```python
n1 = int(input())
t1 = tuple(map(int, input().split()))

n2 = int(input())
t2 = tuple(map(int, input().split()))

nested_tuple = (t1, t2)

print(nested_tuple)
```

```python
Maximum in Tuple


Question 9
In data analysis and programming, finding the largest element in a dataset is a common requirement. Since tuples are immutable sequences in Python, we can still perform operations like finding the maximum element. The task is to read a tuple of integers and print its maximum value.
```
