```python
Union of Two Sets

Python
Question 1
You are tasked with writing a program that reads two sets of integers and computes their union. The union of two sets contains all unique elements that appear in either set (without duplicates). Your program should read the sizes and elements of the sets, then output the union in set format.
```

```python
n = int(input())
set1 = set(map(int, input().split()))

m = int(input())
set2 = set(map(int, input().split()))

result = set1.union(set2)

print(result)
```

```python
Symmetric Difference


Question 4
You are required to write a program that reads two sets of integers and finds their symmetric difference. The symmetric difference of two sets contains all elements that belong to either of the sets but not to both. In other words, it excludes the common elements and includes the unique ones from both sets. The result should be displayed in Python set format.
```

```python
n = int(input())
set1 = set(map(int, input().split()))

m = int(input())
set2 = set(map(int, input().split()))

print(set1.symmetric_difference(set2))
```

```python
Superset Check

Question 6
You are required to write a program that reads two sets of integers and checks if the first set is a superset of the second set.
 A set A is considered a superset of set B if every element of B is also present in A. Your program should output "Yes" if the first set is a superset of the second, and "No" otherwise.
```

```python
n = int(input())
set1 = set(map(int, input().split()))

m = int(input())
set2 = set(map(int, input().split()))

if set1.issuperset(set2):
    print("Yes")
else:
    print("No")
```

```python
Disjoint Sets


Question 7
You are required to write a program that reads two sets of integers and checks if they are disjoint.
 Two sets are said to be disjoint if they have no elements in common. Your program should output "Yes" if the sets are disjoint, otherwise "No".
```

```python
n = int(input())
set1 = set(map(int, input().split()))

m = int(input())
set2 = set(map(int, input().split()))

if set1.isdisjoint(set2):
    print("Yes")
else:
    print("No")
```

```python
Add Element to Set

Question 8
You are required to write a program that reads a set of integers and then adds a new element to it. If the element already exists in the set, it will not be duplicated since sets automatically maintain unique values. Finally, the updated set should be printed in Python set forma
```

```python
Remove Element
Question 9
You are required to write a program that reads a set of integers and then removes a specified element from it.
 If the element exists in the set, it should be removed.
 If the element does not exist, the set remains unchanged.
 Finally, the updated set should be printed in Python set format.
```

```python
n = int(input())

s = set(map(int, input().split()))

x = int(input())

s.discard(x)

print(s)
```
